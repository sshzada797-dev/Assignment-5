```python
class Student:
    def __init__(self, name, roll_number):
        self.__name = name            # Private variable
        self.__roll_number = roll_number  # Private variable

    # Getter for name
    def get_name(self):
        return self.__name

    # Setter for name
    def set_name(self, name):
        self.__name = name

    # Getter for roll_number
    def get_roll_number(self):
        return self.__roll_number

    # Setter for roll_number
    def set_roll_number(self, roll_number):
        self.__roll_number = roll_number

# Testing
s = Student("Abdul Raheem", 786)
print(s.get_name(), s.get_roll_number())
s.set_name("Abdul Raheem")
print(s.get_name())
```

```python
class BankAccount:
    def __init__(self, initial_balance=0):
        self.__balance = initial_balance  # Private variable

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if self.__balance - amount >= 0:
            self.__balance -= amount
        else:
            print("Withdrawing amount exceeds current balance!")

    def get_balance(self):
        return self.__balance

# Testing
account = BankAccount(500)
account.withdraw(600)  # Error message trigger karega
```

```python
class Employee:
    def __init__(self, salary):
        self.set_salary(salary)

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary < 0:
            print("Invalid Salary")
        else:
            self.__salary = salary

# Testing
emp = Employee(50000)
emp.set_salary(-1000)  # Prints: Invalid Salary
```

```python
class Animal:
    def eat(self):
        return "Eating food..."

class Mammal(Animal):
    def walk(self):
        return "Walking on legs..."

class Dog(Mammal):
    def bark(self):
        return "Woof! Woof!"

# Testing
puppy = Dog()
print(puppy.eat())  # Parent Class 1 feature
print(puppy.walk()) # Parent Class 2 feature
print(puppy.bark()) # Own feature
```

```python
class Shape:
    def info(self):
        return "This is a geometric shape"

class Circle(Shape):
    def draw(self):
        return "Drawing a Circle"

class Square(Shape):
    def draw(self):
        return "Drawing a Square"

# Testing
c = Circle()
s = Square()
print(c.info(), "-", c.draw())
print(s.info(), "-", s.draw())
```

```python
class Cat:
    def make_sound(self):
        return "Meow"

class Dog:
    def make_sound(self):
        return "Woof"

# Polymorphic Function
def make_sound(animal_object):
    print(animal_object.make_sound())

# Testing
make_sound(Cat())
make_sound(Dog())
```

```python
class Book:
    def __init__(self, pages):
        self.pages = pages

    # Overloading '+' operator
    def __add__(self, other):
        return self.pages + other.pages

# Testing
b1 = Book(200)
b2 = Book(150)
total_pages = b1 + b2  # Calls b1.__add__(b2)
print("Total Pages:", total_pages)
```

```python
from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass

class Laptop(Computer):
    def process(self):
        print("Laptop is processing data...")

class Desktop(Computer):
    def process(self):
        print("Desktop is processing data...")

# Testing
l = Laptop()
d = Desktop()
l.process()
d.process()
```

```python
from abc import ABC, abstractmethod

class Car(ABC):
    # Concrete Method (Normal function body ke sath)
    def start_engine(self):
        print("Engine started... Brum brum!")

    # Abstract Method (Bina kisi body ke)
    @abstractmethod
    def drive(self):
        pass

class SportsCar(Car):
    def drive(self):
        print("Driving sports car at high speed!")

# Testing
car = SportsCar()
car.start_engine()
car.drive()
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def process_payment(self, amount):
        pass

class CreditCardPayment(Payment):
    def process_payment(self, amount):
        print(f"Paid {amount} using Credit Card.")

class UPIPayment(Payment):
    def process_payment(self, amount):
        print(f"Paid {amount} using UPI.")

# Testing
p1 = CreditCardPayment()
p2 = UPIPayment()
p1.process_payment(1500)
p2.process_payment(500)
```
